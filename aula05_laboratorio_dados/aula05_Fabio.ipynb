## Questão 1 — `groupby` simples (1,0 ponto)

Calcule, a partir de `df` já limpo (ver preparação padrão):

a) a nota média por `turno`;
b) o número de estudantes (`size`) por `modalidade`;
c) explique, em uma frase, a diferença entre usar `.size()` e usar
   `.count()` sobre a coluna `nota_final` dentro de um `groupby`.

````python
# ESCREVA SEU CÓDIGO AQUI
#a) 
media_por_turno = df.groupby('turno')['nota_final'].mean()
print('a) ----------------')
print(media_por_turno)
#b) 
total_linhas_modalidade = df.groupby('modalidade').size()
print('b) ----------------')
print(total_linhas_modalidade)

#c)
medida_size = df.groupby('modalidade')['nota_final'].size()
medida_count = df.groupby('modalidade')['nota_final'].count()
print('c) ----------------------')
print(f'media usando size{medida_size} e usando count{medida_count}')

#df.info()
````

_Sua resposta escrita:_

a)  usando o comando groupby e especificando a coluna turno entre parenteses e nota_final entre colchetes com o ponto mean no final,
    eu terei a média da nota_final agrupada pelo turno dos alunos.

b) usando o comando groupby e passando a coluna modalidade ponto size, o programa irá retornar o total de linhas que eu possuo na minha coluna modalidade, sem excluir as linhas nulas.

## Questão 3 — `agg` com named aggregation (1,5 ponto)

Usando `groupby("campus")` e **named aggregation** (a forma
`nome=("coluna", "função")`), produza uma tabela com:

- `n_estudantes` — contagem de matrículas;
- `nota_media` — média de `nota_final`;
- `nota_desvio` — desvio padrão de `nota_final`;
- `taxa_aprovacao` — percentual de `situacao == "Aprovado"` (dica: uma
  função `lambda` que calcula a média de uma comparação booleana, vezes 100).

Ordene o resultado por `taxa_aprovacao`, do maior para o menor.

````python
# ESCREVA SEU CÓDIGO AQUI
df.groupby('campus').agg(
n_estudantes = ("matricula", "count"),
nota_media = ("nota_final","mean"),
nota_desvio = ("nota_final", "std"), 
taxa_aprovacao = ("situacao", lambda s: (s=="Aprovado").mean() * 100),
).round(2).sort_values('taxa_aprovacao',ascending=False)


````

_Sua resposta escrita:_

utilizando a função named aggregation é possivel renomear as colunas com suas respectivas funções de agregação, destaque para a função lambda na named_aggregation 'taxa_aprovacao' no qual utiliza lambda que cria uma series com valores de situação e checa se são aprovados e realiza a media dos aprovados multiplicando por 100 para ter o resultado em %.

## Questão 8 — `apply(axis=1)` (1,0 ponto)

Escreva uma função `classificar_carga(linha)` que recebe uma linha de
`df_completo` e devolve:

- `"Curso longo"` se `carga_horaria >= 1800`;
- `"Curso médio"` se `1200 <= carga_horaria < 1800`;
- `"Curso curto"` caso contrário.

Aplique com `df_completo.apply(classificar_carga, axis=1)`, salve em uma
coluna `porte_curso` e mostre `value_counts()`.

````python
# ESCREVA SEU CÓDIGO AQUI
def classificar_carga(linha):
    if  linha['carga_horaria'] >= 1800:
        return 'Curso Longo'
    if linha['carga_horaria'] <= 1800 and linha['carga_horaria'] >= 1200:
        return 'Curso Medio'
    if linha['carga_horaria'] < 1200:    
        return 'Curso curto'
    return "Atencao" 

df['porte_curso'] = df_completo.apply(classificar_carga,axis=1)
df['porte_curso'].value_counts()

````

_Sua resposta escrita:_
Com axis=1, cada chamada de classificar recebe uma linha inteira (uma Series com todas as colunas), o que permite combinar várias colunas numa mesma regra. É a ferramenta certa quando a lógica é condicional demais para caber em uma única expressão vetorizada — mas é também a forma mais lenta de operar em um DataFrame, porque o Pandas monta uma Series por linha antes de cada chamada.


## Questão 9 — Method chaining (1,5 ponto)

Em **uma única cadeia de métodos** (um só `resultado = (...)`, sem
variáveis intermediárias), a partir de `df_completo`:

1. descarte estudantes com `situacao == "Trancado"` (`query`);
2. crie a coluna booleana `aprovado` (`assign`);
3. agrupe por `regiao` e `eixo_tecnologico`;
4. calcule `n` (contagem) e `taxa_aprovacao` (média de `aprovado`) com
   named aggregation;
5. mantenha só grupos com `n >= 10` (`query`);
6. ordene por `taxa_aprovacao` decrescente.

Mostre as 5 primeiras linhas do resultado.

````python
# ESCREVA SEU CÓDIGO AQUI
resultado = (
    df_completo
    .query("situacao != 'Trancado' ")
    .groupby(['regiao','eixo_tecnologico'])
    .agg(n=('matricula','count'), taxa_aprovacao=('aprovado','mean'))
    .query('n >= 10')
    .sort_values('taxa_aprovacao', ascending=False)   
)
resultado.head()
````

_Sua resposta escrita:_

ao utilizar "resultado =  (   ) , os parênteses externos permitem quebrar a expressão em varias linhas sem usar \. cada método devolve um novo objeto, sobre o qual o próximo metodo é chamado, realiza a leitura de cima para baixo que corresponde exatamente à ordem das transformações: filtra, agrupa, agrega, filtra de novo e por fim ordena de forma decrescente.



## Questão 10 — Uma pergunta respondida com dados (1,0 ponto)

> **Estudantes de cursos com carga horária maior têm nota média mais
> alta?**

a) Use `pd.cut()` para criar uma coluna `carga_faixa` a partir de
   `carga_horaria`, com faixas `[0, 1200]`, `(1200, 1800]` e
   `(1800, 3000]`, rotuladas `"Até 1200h"`, `"1201–1800h"` e
   `"Acima de 1800h"`.
b) Agrupe por `carga_faixa` e calcule a nota média (use
   `observed=True` no `groupby` para evitar aviso).
c) Escreva, em uma célula de texto, o que os números mostram — e **uma
   limitação** dessa conclusão (pense em quais outras variáveis diferem
   entre cursos curtos e longos).

````python
# ESCREVA SEU CÓDIGO AQUI
#a) 
df_completo['carga_faixa'] = pd.cut(df_completo['carga_horaria'], bins=[0,1200,1800,3000], labels=['Até 1200h','1201-1800','Acima de 1800h'],)
df_completo.head()

#b)
resultadoCarga = (
    df_completo
    .groupby("carga_faixa",observed=True)["nota_final"]
    .mean()
    .round(2)
)
print(resultadoCarga)
````

_Sua resposta escrita:_

a) metodo pd.cut() é bem utilizado para dividir uma serie em intervalos com seus parametros bins e labels, bins é responsavel por criar as faixas de intervalos, e labels por rotular essas faixas, então ao realizar a aplicação de pd.cut na coluna carga horaria é possivel dividir nas faixas de cursos de até 1200h, entre 1201-1800 e acima de 1800. 


b) utilizando method chaining é possivel realizar a aplicação da questão agrupar o df_completo por carga_faixa e aplicando a média da nota final, descobrindo a média das notas por carga horaria dos cursos.


c) os numeros mostram a média das notas finais dos alunos a partir da carga horaria do curso ocupado, realizar uma analise só através da média não ajuda a ter uma conclusão concreta sobre os dados, poderiamos ter outras variaveis como por exemplo total de alunos por carga horaria, taxa de aprovação de alunos por carga horaria, mediana , desvio padrão para compreender a variabilidade das notas diante do agrupamento por carga horaria, poderiamos utilizar a variavel campus para verificar se são todos os campus que possuem cursos em todas as faixas de carga horaria.
