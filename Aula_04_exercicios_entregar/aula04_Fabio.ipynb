## Questão 8 — Coluna derivada com NumPy (1,0 ponto)

Crie a coluna `faixa` com `np.where`, segundo a regra:

- `nota_final` >= 85 → `"Excelente"`
- 70 <= `nota_final` < 85 → `"Bom"`
- 60 <= `nota_final` < 70 → `"Regular"`
- `nota_final` < 60 → `"Insuficiente"`

**Cuidado:** quem está sem nota deve ficar com `NaN` em `faixa`, e não cair
na última categoria. Mostre `df["faixa"].value_counts(dropna=False)`.

```python
# ESCREVA SEU CÓDIGO AQUI

df_filtros = df.copy()
condicao = np.where(
                        (df_filtros['nota_final'] >= 85) , "Excelente", 
                            np.where(
                                    (df_filtros['nota_final'] < 85) & (df_filtros['nota_final'] >= 70), "Bom",
                                np.where(
                                       (df_filtros['nota_final'] < 70) & (df_filtros['nota_final'] >= 60), "Regular",
                                    np.where(
                                        (df_filtros['nota_final'] < 60), "Irregular",np.nan
                                        )
                                    )
                                )
                            )
                        
                        
df_filtros['faixa'] = condicao
df_filtros["faixa"] = df_filtros["faixa"].mask(df_filtros["nota_final"].isna())
df_filtros["faixa"].value_counts(dropna=False)
```

_Sua resposta escrita:_

para criar a coluna chamada faixa utilizando np.where() foi necessario criar um np.where aninhado para colocar todas as condições , inclusive necessario aplicar NaN para valores NaN aplicando a função isna() do pandas, ao final utilizando value_counts() da para verificar os subgrupos de cada faixa por nota, importante ressaltar que foi necessário criar um novo dataframe com a funcao .copy() para evitar o SettingWithCopyWarning indicando a indexação encadeada.
